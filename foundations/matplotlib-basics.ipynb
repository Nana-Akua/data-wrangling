**Making figures with NumPy arrays**

```python
#create some data
x = np.linspace(0, 10, 100)
x[:10]
```

```python
#line plot 
fig, ax = plt.subplots()
ax.plot(x, x**2);
```

```python
#scatterplot 
fig, ax=plt.subplots()
ax.scatter(x, np.exp(x));
```

```python
#plot from dictionary
nut_butter_prices = {"almond": 10, 
                     "peanut": 8,
                     "cashew": 12}
fig, ax=plt.subplots()
ax.bar(nut_butter_prices.keys(), nut_butter_prices.values())
ax.set(title="Dan's Nut Butter Store")
```

```python
#histograms
x = np.random.randn (1000) 
fig, ax=plt.subplots()
ax.hist(x);
```

```python
#mutlitple plots with subplots 
#option 1 **
fig, ((ax1, ax2),(ax3,ax4)) = plt.subplots(nrows=2,
                                          ncols=2,
                                          figsize=(10,5))
#plot to each axiz
ax1.plot(x, x/2);
ax2.scatter(np.random.random(10), np.random.random(10))
ax3.bar(nut_butter_prices.keys(), nut_butter_prices.values())
ax4.hist(np.random.randn(1000))
```

```python
#mutlitple plots with subplots 
#option 2
fig, ax = plt.subplots(nrows=2, ncols=2, figsize=(10,5))
#plot to each different index
ax[0,0].plot(x, x/2);
ax[0,1].scatter(np.random.random(10), np.random.random(10));
ax[1,0].bar(nut_butter_prices.keys(), nut_butter_prices.values());
ax[1,1,].hist(np.random.randn(1000));
```

*Plotting from Panadas DataFrame*

```python
car_sales = pd.read_csv("car-sales.csv")
car_sales
```

```python
ts = pd.Series(np.random.randn(100), 
               index=pd.date_range("1/1/2020", periods=100))
ts.cumsum()
ts.plot()
```

```python
#plotting a dataframe 
#removing unwanted content from the column
car_sales["Price"] = car_sales["Price"].str.replace(r"[\$\,\.]", '', regex=True)
car_sales
```

```python
#remove the last 2 zeros 
car_sales["Price"] = car_sales["Price"].str[:-2]
car_sales
```

```python
#creating a date column 
#periods is how many dates you want  relative to the date 
#default counter type is days
car_sales["Sale Date"] = pd.date_range("1/1/2020", periods=len(car_sales))
car_sales
```

```python
#Turn price into int use .astype()
```

```python
#make a total column 
#answer: How much we have made from sellinng all of our cars
car_sales["Total"] = car_sales["Price"].astype(int).cumsum()
car_sales
```

```python
#plot total sales 
car_sales.plot(x="Sale Date", y="Total")
```

```python
#plot total sales scatter 
car_sales["Price"] = car_sales["Price"].astype(int)
car_sales.plot(x="Odometer (KM)", y="Total",kind="scatter")
```
