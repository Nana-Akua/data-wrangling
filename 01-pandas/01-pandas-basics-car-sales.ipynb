**Creating new data:**

```python
#new column from series 
#The series length does not need to be equal to the index of the datafram 
seats_column = pd.Series([5, 5, 5, 5, 5, 5])

#making a new column
car_sales["Seats"] = seats_column
car_sales
```

```python
#filling a missing value 
car_sales["Seats"] = car_sales["Seats"].fillna(5)
```

```python
car_sales
```

```python
#new column from a list 
#to use a list the list must match the exact number of rows or it would give you an error 
fuel_economy = [7.5, 4, 23, 5, 6, 5, 8, 4, 6, 5]
car_sales["Fuel"] = fuel_economy
```

```python
car_sales
```

```python
#new column  from existing columns
#essentially treat columns like vairables
car_sales["Total fuel used (L)"] = car_sales["Odometer (KM)"]/100 * car_sales["Fuel"]
```

```python
car_sales
```

```python
#create a colums from a single value 
car_sales["Numnber of wheels"] = 4
car_sales
```

```python
#set a column equal to a state
car_sales["Passed road safety"] = True 
car_sales
```

```python
car_sales["Total fuel used"] = 3
```

```python
#removing columns 
car_sales = car_sales.drop("Total fuel used", axis=1)
```

```python
car_sales
```

```python
#shuffling data 
car_sales_shuffle = car_sales.sample(frac=1)
```

```python
car_sales_shuffle
```

```python
#shuffling a percentage of your data 
#20% of the data is shuffled
car_sales_shuffle = car_sales.sample(frac = 0.2)
```

```python
car_sales_shuffle
```

```python
#want to restore the orginal indext
car_sales_shuffle = car_sales_shuffle.reset_index(drop=True)
car_sales_shuffle
```
