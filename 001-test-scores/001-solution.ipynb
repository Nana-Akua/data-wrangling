#import libraries 
import pandas as pd 
import numpy as np

#random number generator
g = np.random.default_rng(0)

#random number series  
test_scores = pd.Series(g.integers(0, 101, 10))

#setting index of series to monts
test_scores.index = ["Sept", "Oct", "Nov", "Dec", "Jan", "Feb", "Mar", "Apr", "May", "Jun"]

#print test scores
print(test_scores)

**Analysis Questions:**\
q1: What is the student average test score for the entire year?\
q2: What is the student's average test score during the first 5 months of the year?\
q3: What is the student average test score during the last 5 months of the year?\
q4: Did the student improve their performance in the sencond half? If so by how much?\
q5: In which month did this student get their highest score\
q6: What were this students five highest scores?

```python
#Question 1
yearly_avg = test_scores.mean()
print(f'Yearly average: {yearly_avg}')
```

```python
#Question 2
first_sem_avg = test_scores[0:5].mean()
print(f'First semester average: {first_sem_avg}')
```

```python
#Question 3
sec_sem_avg = test_scores[5:].mean()
print(f'Second semester average: {sec_sem_avg}')
```

```python
#Question 4
performance_difference = sec_sem_avg - first_sem_avg
#performance comparator
if sec_sem_avg > first_sem_avg:
    print(f'Student improved their performance by: {performance_difference.round(2)}')
elif sec_sem_avg < first_sem_avg:
    print(f'Student performance reduced by: {performance_difference.round(2) * -1} put them in tutoring!')
else:
    print("performance did not change")
```

```python
#Question 5
highest_month = test_scores.idxmax()
max_test_score = test_scores.max()
print(f'{highest_month} was the month you scored the highest with a: {max_test_score}')
```

```python
#Question 6
max_test_scores = test_scores.sort_values(ascending=False)
print(f' You're highest scores were: {max_test_scores.head(5).values}')
```
